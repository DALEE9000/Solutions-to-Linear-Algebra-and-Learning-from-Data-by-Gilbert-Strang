```python
import numpy as np
```

```python
mylist = []
abclist = []
interval = 0.01

for a in np.arange(0,4,interval):
    for b in np.arange(0,4,interval):
        for c in np.arange(0,4,interval):
            abclist.append((a,b,c))
            row1 = np.abs(3 - a) + np.abs(0 - b)
            row2 = np.abs(4 - c*a) + np.abs(5 - c*b)
            mylist.append(max(row1, row2))

min(mylist)
```
