I.5.2. Draw unit vectors $\bm{u}$ and $\bm{v}$ that are not orthogonal. Show that $\bm{w} = \bm{v} - \bm{u}(\bm{u}^{T} \bm{v})$ is orthogonal to $\bm{u}$ (and add $\bm{w}$ to your picture).

```python
plt.figure(figsize=(4,4))

t = np.arange(0,1/np.sqrt(2),0.01)

plt.hlines(0,0,1, linewidth=2, color=FG)
plt.vlines(0,0,1/np.sqrt(2), linewidth=2, color=C.orange)
plt.plot(t,t, linewidth=2, color=FG)
plt.arrow(0,0,1,0, color=FG, shape='full', lw=0, length_includes_head=True, head_width=.05)
plt.arrow(0,0,1/np.sqrt(2),1/np.sqrt(2), color=FG, shape='full', lw=0, length_includes_head=True, head_width=.05)
plt.arrow(0,0,0,1/np.sqrt(2), color=C.orange, shape='full', lw=0, length_includes_head=True, head_width=.05)

plt.text(0.6, 0.75, '$v = (1/\\sqrt{2}, 1/\\sqrt{2})$', fontsize=14)
plt.text(1.05, 0, '$u = (1, 0)$', fontsize=14)
plt.text(-0.3, 0.75, '$w = (0, 1/\\sqrt{2})$', fontsize=14)

plt.xlim(-0.5, 1.7)
plt.ylim(-0.1, 1)
plt.minorticks_off()

savefig(plt.gcf(), 'partiseci.5ex2.eps', bbox_inches=None)
plt.show()
```

I.5.3. Draw any two vectors $\bm{u}$ and $\bm{v}$ out from the origin $(0, 0)$. Complete two more sides to make a parallelogram with diagonals $\bm{w = u + v}$ and $\bm{z = u - v}$. Show that $\bm{w^{T}w + z^{T}z}$ is equal to $\bm{2u^{T} + 2v^{T}v}$.

```python
plt.figure(figsize=(4,4))

t = np.arange(0,1,0.01)

plt.hlines(0,0,1, linewidth=1, color=FG)
plt.vlines(0,0,1, linewidth=1, color=FG)
plt.hlines(1,0,1, linewidth=1, color=FG)
plt.vlines(1,0,1, linewidth=1, color=FG)
plt.plot(t,t, linewidth=1, color=FG)
plt.plot(t,-t + 1, linewidth=1, color=FG)

plt.arrow(0,0,1,0, color=FG, shape='full', lw=0, length_includes_head=True, head_width=.05)
plt.arrow(0,0,0,1, color=FG, shape='full', lw=0, length_includes_head=True, head_width=.05)
plt.arrow(0,1,1,0, color=FG, shape='full', lw=0, length_includes_head=True, head_width=.05)
plt.arrow(1,0,0,1, color=FG, shape='full', lw=0, length_includes_head=True, head_width=.05)
plt.arrow(0,0,1,1, color=FG, shape='full', lw=0, length_includes_head=True, head_width=.05)
plt.arrow(0,1,1,-1, color=FG, shape='full', lw=0, length_includes_head=True, head_width=.05)

plt.text(0, 1.05, '$v$', fontsize=14)
plt.text(1.05, 0, '$u$', fontsize=14)
plt.text(0.6, 0.7, '$w$', fontsize=14)
plt.text(0.6, 0.25, '$z$', fontsize=14)

plt.xlim(-0.1, 1.2)
plt.ylim(-0.1, 1.2)
plt.minorticks_off()

savefig(plt.gcf(), 'partiseci.5ex3.eps', bbox_inches=None)
plt.show()
```
