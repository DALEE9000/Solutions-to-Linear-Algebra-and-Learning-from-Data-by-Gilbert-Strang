I.1.6. If three corners of a parallelogram are $(1, 1)$, $(4, 2)$, and $(1, 3)$, what are all three of the possible fourth corners? Draw two of them.

```python
exes = [1, 4, 1, -2, 4, 4]
whys = [1, 2, 3, 2, 0, 4] 

t1 = np.arange(-2,1,0.01)
t2 = np.arange(1,4,0.01)

y1 = (1/3)*t1 + 8/3
y2 = (-1/3)*t1 + 4/3
y3 = (1/3)*t2 + 2/3
y4 = (-1/3)*t2 + 10/3
y5 = (-1/3)*t2 + 4/3
y6 = (1/3)*t2 + 8/3

plt.figure(figsize=(6,4))
plt.scatter(exes, whys, s=40, facecolors=FG, edgecolors=FG, zorder=2, clip_on=False)
plt.plot(t1, y1, linewidth=2, color=C.red, zorder=0)
plt.plot(t1, y2, linewidth=2, color=C.red, zorder=0)
plt.plot(t2, y3, linewidth=2, color=C.red, zorder=0)
plt.plot(t2, y4, linewidth=2, color=C.red, zorder=0)
plt.plot(t2, y4, linewidth=2, linestyle='dotted', color=C.green, zorder=0)
plt.plot(t2, y5, linewidth=2, linestyle='dotted', color=C.green, zorder=0)
plt.vlines(1, 1, 3, linewidth=2, linestyle='dotted', color=C.green, zorder=0)
plt.vlines(4, 0, 2, linewidth=2, linestyle='dotted', color=C.green, zorder=0)
plt.vlines(4, 2, 4, linewidth=2, linestyle='dashdot', color=C.blue, zorder=0)
plt.vlines(1, 1, 3, linewidth=2, linestyle='dashdot', color=C.blue, zorder=0)
plt.plot(t2, y6, linewidth=2, linestyle='dashdot', color=C.blue, zorder=0)
plt.plot(t2, y3, linewidth=2, linestyle='dashdot', color=C.blue, zorder=0)
plt.xlim(-3,5)
plt.ylim(-1,5)
plt.minorticks_off()
savefig(plt.gcf(), 'partiseci.1ex6.eps', bbox_inches=None)
plt.show()
```
