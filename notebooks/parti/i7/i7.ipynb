I.7.21. Draw the tilted ellipse $x^2 + xy + y^2 = 1$ and find the half-lengths of its axes from the eigenvalues of the corresponding matrix $S$.

```python
plt.figure(figsize=(6,6))

x = np.arange(0,1,0.01)
y = np.arange(0,0.56,0.01)

t = np.arange(0, 2*np.pi, 0.01)
t_rot = np.pi/4

Ell = np.array([(1/np.sqrt(3/2))*np.cos(t), (1/np.sqrt(1/2))*np.sin(t)])
R_rot = np.array([[np.cos(t_rot) , - np.sin(t_rot)],[np.sin(t_rot) , np.cos(t_rot)]])  
     #2-D rotation matrix

Ell_rot = np.zeros((2,Ell.shape[1]))
for i in range(Ell.shape[1]):
    Ell_rot[:,i] = np.dot(R_rot,Ell[:,i])

plt.plot(Ell_rot[0,:], Ell_rot[1,:], linewidth=2, color=FG)
plt.scatter(0, 0, s=40, facecolors=FG, edgecolors=FG, zorder=2, clip_on=False)

plt.plot(x, -x, linewidth=2, linestyle='dashed', color=FG)
plt.arrow(0,0,0.56,0.56, color=FG, shape='full', lw=0, length_includes_head=True, head_width=.1)
plt.plot(y, y, linewidth=2, linestyle='dashed', color=FG)
plt.arrow(0,0,1,-1, color=FG, shape='full', lw=0, length_includes_head=True, head_width=.1)

plt.text(0, 0.35, '$1/\\sqrt{\\lambda_1}$', fontsize=16, ha='center')
plt.text(0.3, -0.7, '$1/\\sqrt{\\lambda_2}$', fontsize=16, ha='center')

plt.xlim(-1.5, 1.5)
plt.ylim(-1.5, 1.5)
plt.minorticks_off()

savefig(plt.gcf(), 'partiseci.7ex21.eps', bbox_inches=None)
plt.show()
```
